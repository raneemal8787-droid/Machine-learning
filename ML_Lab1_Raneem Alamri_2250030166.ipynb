Task6

```python
marks = [75, 82, 91, 68, 88]
print("Marks:", marks)
print("First mark:", marks[0])
print("Third mark:", marks[2])
print("Number of marks:", len(marks))
print("Total:", sum(marks))
print("Highest:", max(marks))
print("Lowest:", min(marks))
print("Average:", sum(marks) / len(marks))
```
