```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style="whitegrid")
```

```python

df = pd.read_csv('Chocolate_Sales.csv.csv')

display(df.head())


print("Shape (Rows, Columns):", df.shape)


print("\nData Types:")
print(df.dtypes)


display(df.describe(include='all'))
```

```python

print("Missing values:\n", df.isna().sum())

duplicates_count = df.duplicated().sum()
print(f"\nDuplicate rows count: {duplicates_count}")
if duplicates_count > 0:
    df = df.drop_duplicates()
    print("Duplicates removed.")

df.columns = df.columns.str.strip()
print("\nColumns:", df.columns.tolist())
```

```python

plt.figure(figsize=(10, 4))
sns.histplot(df['Boxes Shipped'], kde=True, color='purple')
plt.title('Distribution of Boxes Shipped')
plt.show()

if 'Revenue' in df.columns:
    if df['Revenue'].dtype == 'object':
        df['Revenue'] = df['Revenue'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False).astype(float)

    plt.figure(figsize=(10, 4))
    sns.histplot(df['Revenue'], kde=True, color='green')
    plt.title('Distribution of Revenue')
    plt.show()
```

```python

if 'Country' in df.columns and 'Revenue' in df.columns:
    plt.figure(figsize=(12, 6))
    sns.barplot(x='Country', y='Revenue', data=df, estimator=np.sum, palette='viridis')
    plt.title('Total Revenue by Country')
    plt.xticks(rotation=45)
    plt.show()

plt.figure(figsize=(8, 6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()
```
